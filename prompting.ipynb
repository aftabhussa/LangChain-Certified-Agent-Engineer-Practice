```python
from langchain.agents import create_agent
from langchain.messages import HumanMessage
agent = create_agent("gpt-5-nano")
question = HumanMessage(content="Write first four alphabets of english language")
response = agent.invoke(
    {'messages':[question]}
    )
print(response['messages'][-1].content)
```

```python
system_prompt = "You are an english teacher, give alphabets with 1 missing letter"

eng_agent = create_agent(
    model="gpt-5-nano",
    system_prompt=system_prompt
)

response=eng_agent.invoke(
    {'messages':[question]}
)

print(response['messages'][-1].content)
```

```python
from langchain.agents import create_agent
from pydantic import BaseModel
from langchain.messages import HumanMessage

class StructureOutput(BaseModel):
    city : str
    location : str
    population : int

agent = create_agent(
    model = 'gpt-5-nano',
    system_prompt="You are an expert demographer and give answer from around the world",
    response_format = StructureOutput
)

for token, metadata in agent.stream(
    {
        "messages": [
            HumanMessage(
                content="Tell us about the most populated city in the world"
            )
        ]
    },
    stream_mode="messages"
):
    if token.content:
        print(token.content, end="", flush=True)
```
