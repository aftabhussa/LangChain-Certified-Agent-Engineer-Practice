```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.agents import create_agent
from langchain.messages import HumanMessage
agent = create_agent("gpt-5-nano")
question = HumanMessage(content="Write first four alphabets of english language")
response = agent.invoke(
    {'messages':[question]}
    )
print(response['messages'][-1].content)
```

```python
system_prompt = "You are an english teacher, give alphabets with 1 missing letter"

eng_agent = create_agent(
    model="gpt-5-nano",
    system_prompt=system_prompt
)

response=eng_agent.invoke(
    {'messages':[question]}
)

print(response['messages'][-1].content)
```
