```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from ipywidgets import FileUpload
from IPython.display import display

uploader = FileUpload(accept='.png', multiple=False)
display(uploader)
```

```python
import base64
uploaded_file = uploader.value[0]
read_bytes = bytes(uploaded_file['content'])
img_encode = base64.b64encode(read_bytes).decode('utf-8')
print(len(img_encode))
```

```python
from langchain.messages import HumanMessage
from langchain.agents import create_agent

question = HumanMessage(content=[
    {"type": "text", "text": "Explain this picture"},
    {"type": "image", "base64": img_encode, "mime_type": "image/png" }
])

agent = create_agent(
    model = 'gpt-5-nano',
)

response = agent.invoke(
    {'messages': [question]}
)

print(response['messages'][-1].content)
```

```python
import sounddevice as sd
import io
from scipy.io.wavfile import write
import time
from tqdm import tqdm
```

```python
duration = 5
sample_rate = 44100

audio = sd.rec(int(duration * sample_rate), samplerate=sample_rate, channels=1)

for _ in tqdm(range(duration*10)):
    time.sleep(0.1)
sd.wait()
print("Done")

buf = io.BytesIO()

write(buf, sample_rate, audio)

wav_bytes = buf.getvalue()

aud_64 = base64.b64encode(wav_bytes).decode('utf-8')
```

```python
from langchain.messages import HumanMessage
from langchain.agents import create_agent

question = HumanMessage(content=[
    {"type": "text", "text": "Write script of this audio file"},
    {"type": "audio", "base64": aud_64, "mime_type": "audio/wav" }
])

agent = create_agent(
    model = 'gpt-audio',
)

response = agent.invoke(
    {'messages': [question]}
)

print(response['messages'][-1].content)
```

```python
from ipywidgets import FileUpload
from IPython.display import  display

file_upload = FileUpload(accept='.pdf', multiple=False)
display(file_upload)
```

```python
import base64

file=file_upload.value[0]

file_bytes = bytes(file['content'])
final_file=base64.b64encode(file_bytes).decode('utf-8')
print(len(final_file))
```

```python
multimodal=HumanMessage(content=[
                        {"type": "text", "text": "Analyze the file and give skills mentioned in resume"},
                        {"type": "file", "base64": final_file, "mime_type": "application/pdf"}])

agent = create_agent(
    model="gpt-5-nano"
)
response = agent.invoke(
    {'messages':[multimodal]}
)

print(response['messages'][-1].content)
```
