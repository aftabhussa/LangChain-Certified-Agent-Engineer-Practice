```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.agents import create_agent
agent = create_agent(
    model = "gpt-5-nano"
)
```

```python
from langchain.messages import HumanMessage
question = HumanMessage(content="Hi, My name is Aftab & my hobby is playing cricket")

response = agent.invoke(
    {'messages':[question]}
)
print(response['messages'][-1])

```

```python
question=HumanMessage(content="What is my hobby?")

response=agent.invoke(
    {'messages':[question]}
)
print(response['messages'][-1])
```

```python
from langgraph.checkpoint.memory import InMemorySaver
agent = create_agent(
    model = "gpt-5-nano",
    checkpointer=InMemorySaver()
)
```

```python
from pprint import pprint
question=HumanMessage(content="Hi my name is Aftab and I want to explore the world")
config={"configurable":{"thread_id": "1"}}

response=agent.invoke(
    {'messages':[question]},
    config,
)
pprint(response['messages'])
```

```python
question=HumanMessage(content="What is my name and what i want?")

response=agent.invoke(
    {'messages':[question]},
    config,
)
pprint(response['messages'])
```
