```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.tools import tool

@tool

def square_root(x:float) -> float:
    """calculate the square root of a number"""
    return x ** 0.5
```

```python
@tool("square_root")
def tool1(x:float) -> float:
    """calculate the square root of a number0"""
    return x ** 0.5
```

```python
@tool("square_root", description="calculate the square root of a number")

def tool1(x:float) -> float:
    return x ** 0.5
```

```python
tool1.invoke({"x":125})
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model = "gpt-5-nano",
    tools=[tool1],
    system_prompt="You are an expert mathematician and calculate the square root of any number."
)
```

```python
from langchain.messages import HumanMessage

response = agent.invoke(
    {'messages':HumanMessage(content="calculate the square root of 120")}
)

print(response['messages'][-1].content)
```
