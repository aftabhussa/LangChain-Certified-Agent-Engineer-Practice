```python
from dataclasses import dataclass

@dataclass
class colorcontext:
    favourite_color : str = "Green"
    least_favourit_color: str = "Blue"
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model = "gpt-5-nano",
    context_schema=colorcontext
)
```

```python
from langchain.messages import HumanMessage

response = agent.invoke(
    {"messages":[HumanMessage(content="What is my favourite color?")]},
    context=colorcontext()
)
```

```python
from pprint import pprint
pprint(response["messages"][-1].content)
```

```python
from langchain.tools import tool, ToolRuntime

@tool
def get_favourite_color(runtime: ToolRuntime[colorcontext]) -> str:
    "Get the favourite color of user"
    return runtime.context.favourite_color

def get_least_favourite(runtime: ToolRuntime[colorcontext]) -> str:
    "Get the least favourite color of user"
    return runtime.context.least_favourit_color
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model="gpt-5-nano",
    tools = [get_favourite_color, get_least_favourite],
    context_schema=colorcontext
)
```

```python
from langchain.messages import HumanMessage

response = agent.invoke(
    {"messages":[HumanMessage(content="What is my favourite color")]},
    context=colorcontext(favourite_color="Black")
)
```

```python
from pprint import pprint
pprint(response["messages"][-1].content)
```
