```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.chat_models import init_chat_model
model = init_chat_model(model="gpt-5-nano")
```

```python
response = model.invoke("How many province Pakistan have?")
response.content
```

```python
from pprint import pprint
pprint(response.response_metadata)
```

```python
from langchain_google_genai import GoogleGenerativeAI
model = GoogleGenerativeAI(model="gemini-3.1-flash-lite")
```

```python
response = model.invoke("List 5 big cities in the world")
print(response)
```

```python
from langchain.agents import create_agent
agent = create_agent(model=model)
```

```python
agent = create_agent(model="claude-sonnet-4-5")
```

```python
agent=create_agent("gpt-5-nano")
```

```python
from langchain.messages import HumanMessage
response = agent.invoke(
    {"messages":[HumanMessage(content="Which breakfast is best for diabetic?")]}
)
```

```python
from pprint import pprint
pprint(response)
```

```python
print(response['messages'][-1].content)
```

```python
from langchain.messages import AIMessage

response = agent.invoke(
    {"messages":[HumanMessage(content="Which breakfast is best for diabetic?"),
                 AIMessage(content="Oatmeal and Wholegrain toast is best?"),
                 HumanMessage(content="Tell me how to live a healthy life?")]
    }
)
```

```python
print(response)
```

```python
print(response['messages'][-1].content)
```

```python
for token, metadata in agent.stream(
    {'messages':[HumanMessage(content="write 5 benefits of morning walk")]},
    stream_mode = 'messages'
):
    if token.content:
        print(token.content, end='', flush=True)
```
