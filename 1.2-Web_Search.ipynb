```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model="gpt-5-nano"
)
```

```python
from langchain.messages import HumanMessage
response = agent.invoke(
    {'messages':[HumanMessage(content="What is current weather of New York?")]}
)
print(response)
```
